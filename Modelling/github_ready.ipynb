```
import nbformat

path = "/content/drive/MyDrive/Colab Notebooks/Modelling_SMARD_EnergyProduction_Germany_Monthly.ipynb"

nb = nbformat.read(path, as_version=4)

# --- 1. Clean notebook-level widget metadata ---
if "widgets" in nb["metadata"]:
    del nb["metadata"]["widgets"]

# --- 2. Clean cell-level widget metadata ---
for cell in nb.cells:
    if "metadata" in cell:
        # remove widget-related metadata keys commonly inserted by Colab
        for key in [
            "widgets",
            "widget_state",
            "application/vnd.jupyter.widget-view+json",
            "application/vnd.jupyter.widget-state+json"
        ]:
            if key in cell["metadata"]:
                del cell["metadata"][key]

    # --- 3. Clean output metadata (for code cells with outputs) ---
    if "outputs" in cell:
        for output in cell["outputs"]:
            if "metadata" in output:
                for key in [
                    "widgets",
                    "widget_state",
                    "application/vnd.jupyter.widget-view+json",
                    "application/vnd.jupyter.widget-state+json"
                ]:
                    if key in output["metadata"]:
                        del output["metadata"][key]

nbformat.write(nb, path)

print("✔ All widget metadata removed!")
```
